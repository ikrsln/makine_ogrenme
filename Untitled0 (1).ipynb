```python
import urllib.request, zipfile, io

url = "https://archive.ics.uci.edu/static/public/296/diabetes+130-us+hospitals+for+years+1999-2008.zip"
istek = urllib.request.Request(url, headers={"User-Agent": "Mozilla/5.0"})

with urllib.request.urlopen(istek) as r:
    z = zipfile.ZipFile(io.BytesIO(r.read()))

print(z.namelist())
z.extract("diabetic_data.csv", ".")
```

```python
import warnings
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

warnings.filterwarnings("ignore")

# 1) Veriyi oku ("?" eksik değer demek)
df = pd.read_csv("diabetic_data.csv", na_values="?", low_memory=False)

# 2) weight aralıklarını orta noktaya çevir -> sayısal hedef
agirlik = {"[0-25)": 12.5, "[25-50)": 37.5, "[50-75)": 62.5, "[75-100)": 87.5,
           "[100-125)": 112.5, "[125-150)": 137.5, "[150-175)": 162.5,
           "[175-200)": 187.5, ">200": 200}
hedef = "weight_num"
df[hedef] = df["weight"].map(agirlik)

# 3) Modele girmeyecek sütunlar: readmitted, hedef, ID'ler, tanı kodları
cikar = ["readmitted", "weight", hedef,
         "encounter_id", "patient_nbr", "diag_1", "diag_2", "diag_3"]

X_tum = df.drop(columns=cikar).fillna("Bilinmiyor")
X_tum = pd.get_dummies(X_tum)

# 4) Dolu olanlar: eğitim + doğrulama | Boş olanlar: tahmin edilecekler
dolu = df[hedef].notna()
X_dolu, y_dolu = X_tum[dolu], df.loc[dolu, hedef]
X_bos = X_tum[~dolu]
print(f"Dolu satır: {dolu.sum()} | Boş satır: {(~dolu).sum()}")

X_train, X_val, y_train, y_val = train_test_split(
    X_dolu, y_dolu, test_size=0.2, random_state=42)

# 5) Kıyas: hep ortalamayı tahmin eden model
base_mse = mean_squared_error(y_val, np.full(len(y_val), y_train.mean()))
print(f"Baz çizgi (ortalama) MSE = {base_mse:.2f}")

# 6) Modelleri dene
adaylar = {
    "LinearRegression": (LinearRegression, [None]),
    "Ridge": (Ridge, [0.1, 1, 10, 100, 1000, 5000]),
    "Lasso": (Lasso, [0.01, 0.1, 1, 10]),
}

en_iyi_mse, en_iyi_model, en_iyi_ad = float("inf"), None, None
for ad, (Sinif, alphalar) in adaylar.items():
    print(f"\n--- {ad} ---")
    for alpha in alphalar:
        reg = Sinif() if alpha is None else Sinif(alpha=alpha)
        model = Pipeline([("scaler", StandardScaler()), ("reg", reg)])
        model.fit(X_train, y_train)
        tahmin = model.predict(X_val)
        mse = mean_squared_error(y_val, tahmin)
        r2 = r2_score(y_val, tahmin)
        etiket = "" if alpha is None else f"alpha={alpha}: "
        print(f"{etiket}Doğrulama MSE = {mse:.2f} | R2 = {r2:.3f}")
        if mse < en_iyi_mse:
            en_iyi_mse, en_iyi_model = mse, model
            en_iyi_ad = ad if alpha is None else f"{ad} (alpha={alpha})"

print(f"\nEn iyi model: {en_iyi_ad}, MSE = {en_iyi_mse:.2f} (baz çizgi: {base_mse:.2f})")

# 7) Boş weight değerlerini doldur (0-200 aralığına kırp)
df["weight_imputed"] = ~dolu
df.loc[~dolu, hedef] = np.clip(en_iyi_model.predict(X_bos), 0, 200)

print("\nKalan boş sayısı:", df[hedef].isna().sum())
print("\nGerçek değerler:")
print(df.loc[dolu, hedef].describe())
print("\nTahmin edilen değerler:")
print(df.loc[~dolu, hedef].describe())
```
