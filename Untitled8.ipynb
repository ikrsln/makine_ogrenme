```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

!pip install ucimlrepo -q
from ucimlrepo import fetch_ucirepo

dataset = fetch_ucirepo(id=296)
df = pd.concat([dataset.data.features, dataset.data.targets], axis=1)

df.replace('?', np.nan, inplace=True)

df.head(10)
```

```python
df.tail()
```

```python
df.describe(include='object')
```

```python
df.drop(['weight', 'payer_code', 'examide', 'citoglipton'], axis=1, inplace=True)
```

```python
specialty_mode = df['medical_specialty'].mode()[0]
df.fillna({'medical_specialty': specialty_mode}, inplace=True)

race_mode = df['race'].mode()[0]
df.fillna({'race': race_mode}, inplace=True)

for col in ['diag_1', 'diag_2', 'diag_3']:
    diag_mode = df[col].mode()[0]
    df.fillna({col: diag_mode}, inplace=True)

df.isnull().sum()
```

```python
df['total_visits'] = df['number_outpatient'] + df['number_emergency'] + df['number_inpatient']
df.drop(['number_outpatient', 'number_emergency', 'number_inpatient'], axis=1, inplace=True)

df['readmitted_30d'] = (df['readmitted'] == '<30').astype(int)
df.drop(['readmitted'], axis=1, inplace=True)

df.head()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.select_dtypes(include=[np.number]).corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Sayısal Değişkenlerin Korelasyon Matrisi")
plt.show()
```

```python
df.drop(['diag_1', 'diag_2', 'diag_3', 'medical_specialty'], axis=1, inplace=True)

categorical_cols = ['race', 'gender', 'age', 'max_glu_serum', 'A1Cresult', 'change', 'diabetesMed']

df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
df.describe()
```
